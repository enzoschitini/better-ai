# ID Generator

Esta é uma documentação detalhada e didática da classe `IDGenerator`, projetada para fornecer uma visão clara de suas capacidades e funcionamento interno.

---

## Visão Geral

A classe `IDGenerator` é uma ferramenta utilitária robusta desenvolvida em Python para a criação de identificadores únicos. Em sistemas de software, a escolha do ID correto é crucial: alguns precisam ser ordenáveis no tempo (como logs), outros precisam ser impossíveis de adivinhar (como tokens de recuperação de senha) e outros precisam seguir formatos de negócio específicos (como protocolos de atendimento).

Esta classe centraliza essas necessidades em uma interface simples, utilizando bibliotecas padrão do Python focadas em segurança (`secrets`) e unicidade (`uuid`).

---

## Fluxo de Execução

O fluxo de funcionamento da classe baseia-se no padrão **Static Utility**. Como todos os métodos são decorados com `@classmethod`, você não precisa instanciar a classe (fazer `gen = IDGenerator()`) para usá-la.

1. **Chamada do Método:** O usuário solicita um tipo de ID (ex: `IDGenerator.token()`).
2. **Validação de Entrada:** O método verifica se os parâmetros (como tamanho ou versão) são válidos.
3. **Processamento/Geração:**
    - Para **UUIDs**, consulta o módulo `uuid`.
    - Para **Timestamps**, captura o tempo em nanosegundos (`time.time_ns()`).
    - Para **Custom/Tokens**, utiliza o `secrets` para selecionar caracteres aleatórios de forma criptograficamente segura.
4. **Retorno:** O identificador é formatado como uma `string` e retornado ao chamador.

---

## Tabela de Métodos

| **Método** | **Finalidade Principal** | **Característica Chave** |
| --- | --- | --- |
| `uuid(version)` | Identificação universal única. | Padrão RFC 4122 (128 bits). |
| `timestamp(...)` | IDs que seguem uma ordem cronológica. | Baseado em nanosegundos + sufixo aleatório. |
| `token(len, safe)` | Segurança e autenticação. | Criptograficamente forte e seguro para URLs. |
| `custom(pattern)` | Identificadores com regras de negócio. | Usa máscaras para definir números e letras. |

---

## Arquitetura e Insights

- **Segurança Criptográfica:** O uso de `secrets` em vez de `random` garante que os IDs gerados para tokens não possam ser previstos por ataques de força bruta ou análise estatística.
- **Alta Granularidade:** Ao usar `time.time_ns()` no método de timestamp, a chance de colisão (dois IDs iguais gerados ao mesmo tempo) é drasticamente reduzida, pois a precisão é de bilionésimos de segundo.
- **Flexibilidade por Máscaras:** O método `custom` funciona como um "Mini-Template Engine", permitindo que desenvolvedores criem formatos como `NF-2024-####` sem precisar reescrever lógica de geração aleatória.

## Detalhamento Técnico

## Classe IDGenerator

**Descrição**

Uma classe utilitária de métodos de classe voltada para a geração de strings de identificação. Não requer estado interno e foca na pureza das funções de geração.

**Argumentos**

A classe em si não recebe argumentos de inicialização (é uma classe estática).

**Métodos**

## 1. uuid

**Descrição:** Gera um Identificador Único Universal (UUID). É a melhor escolha para chaves primárias de bancos de dados distribuídos.

**Argumentos:**

- `version` (int): A versão do UUID. Suporta `1` (tempo + MAC address) ou `4` (totalmente aleatório). Padrão: `4`.

**Retornos:**

- `str`: O UUID formatado (ex: `8f3...`).

**Raises:**

- `ValueError`: Se a versão informada for diferente de 1 ou 4.

**Exemplos:**

```bash
IDGenerator.uuid(version=4) # '784a0d9b-2b41-4c12-8877-6f8d92305381'
```

---

## 2. timestamp

**Descrição:** Cria um ID baseado no momento exato da execução. Excelente para logs e sistemas onde a ordem de criação importa.

**Argumentos:**

- `prefix` (str): Texto inicial.
- `separator` (str): Caractere de ligação.
- `as_hex` (bool): Se `True`, converte o tempo para base hexadecimal (mais curto).
- `suffix_len` (int): Quantidade de caracteres aleatórios extras ao final para evitar colisões.

**Retornos:**

- `str`: String temporal formatada.

**Raises:**

- `ValueError`: Se `suffix_len` for negativo.

**Exemplos:**

```bash
IDGenerator.timestamp(prefix="PEDIDO", separator="-", as_hex=True) 
# 'PEDIDO-18f74d0a2bc5f1a39aB3'
```

---

## 3. token

**Descrição:** Gera uma sequência aleatória segura. Recomendado para senhas temporárias, chaves de API e tokens de sessão.

**Argumentos:**

- `length` (int): Comprimento da string.
- `url_safe` (bool): Se `True`, remove caracteres que podem causar erro em URLs (como `+` ou `/`).

**Retornos:**

- `str`: Token aleatório.

**Raises:**

- `ValueError`: Se `length` for menor ou igual a zero.

**Exemplos:**

```bash
IDGenerator.token(length=12, url_safe=True) # 'kL9_zP2mNx8A'
```

---

## 4. custom

**Descrição:** Gera um ID seguindo um modelo visual (máscara) definido pelo usuário.

**Argumentos:**

- `pattern` (str): String contendo os marcadores `#` (número), `?` (letra) ou  (ambos).

**Retornos:**

- `str`: String formatada conforme o padrão.

**Raises:**

- `ValueError`: Se o padrão estiver vazio.

**Exemplos:**

```bash
IDGenerator.custom("PLA-####-??") # 'PLA-5521-XW'
```
