# 🎨 Acquarello

**Transforme ideias em imagens.**
Da palavra à imagem, ou de uma imagem a outra — sempre no seu estilo.

O **Acquarello** é uma aplicação de geração de imagem por IA, construída em **Streamlit**, que coloca o poder dos modelos generativos na mão de qualquer pessoa — sem código, sem complicação. Descreva o que imagina e receba uma imagem; ou envie uma imagem existente e reinvente-a. Em ambos os casos, você define o **estilo visual** direto na barra lateral e vê o resultado ganhar a sua cara.

### ✨ Destaques

- 🖋️ **Texto → Imagem** — descreva uma cena e deixe a IA criá-la.
- 🖼️ **Imagem → Imagem** — parta de uma imagem e transforme-a em algo novo.
- 🎚️ **Estilo sob controle** — escolha e ajuste o estilo pela barra lateral.
- ⚡ **Simples de usar** — interface intuitiva, pensada para qualquer usuário.

## Contexto

Modelos generativos de imagem estão entre os avanços mais impressionantes da IA recente — mas, na prática, ainda intimidam quem não é técnico: exigem configuração, prompts complexos e ferramentas dispersas. O **Acquarello** nasce para resolver isso, reunindo geração por texto, geração por imagem e controle de estilo em **uma única interface acessível**. A proposta é aproximar qualquer pessoa da criação visual com IA — do curioso ao criativo — mostrando que essa tecnologia pode ser poderosa *e* simples ao mesmo tempo.
