O **Gerador de Conteúdo** é uma aplicação de IA, construída em **Streamlit**, que produz conteúdo textual alinhado à marca com agilidade e consistência. Diferente de um chat genérico, ele é conectado a uma **base de conhecimento no Pinecone**, onde diversos documentos e catálogos do **Grupo O Boticário** foram embedados. Assim, cada texto gerado é **fundamentado na identidade real da marca** — a IA entende o padrão e o mantém em todo post que escreve.

Você descreve o **objetivo**, adiciona **requisitos**, ajusta as **configurações** e faz um **pedido**. Em segundos, recebe múltiplos posts prontos para publicar.

### ✨ Destaques

- 🧠 **Grounded na marca (RAG + Pinecone)** — busca trechos reais dos catálogos antes de escrever, reduzindo alucinação e mantendo a identidade.
- 🎛️ **Personalização profunda** — objetivo, requisitos extras, quantidade, profundidade de contexto, tamanho, modelo de LLM e idioma.
- 📦 **Múltiplos posts por pedido** — gere um lote de variações de uma só vez.
- ⚡ **Rápido e sem código** — do briefing ao conteúdo em poucos cliques.

## Contexto

Equipes de marketing precisam produzir muito conteúdo, rápido — mas **na voz certa da marca**. Ferramentas genéricas de IA resolvem a velocidade, porém escorregam no essencial: saem do tom, ignoram o portfólio real e às vezes inventam informação.

O **Gerador de Conteúdo** nasce para fechar essa lacuna. Ao ancorar a geração em uma **base vetorial (Pinecone)** alimentada com os catálogos do OBoticário, ele une o melhor dos dois mundos: a **agilidade** de um LLM com a **consistência** de quem realmente conhece a marca. O resultado é conteúdo escalável, coerente e pronto para publicar.

## Como funciona

O projeto segue uma arquitetura **RAG (Retrieval-Augmented Generation)**:

![image](../images/content_generator_schema.png)

1. **Indexação** — os documentos do OBoticário são transformados em embeddings e armazenados no Pinecone.
2. **Recuperação** — a cada pedido, o sistema busca os trechos mais relevantes da base (controlado por *Máximo de resultados*).
3. **Geração** — o LLM escreve os posts usando esse contexto + as suas configurações, respeitando tom, tamanho e idioma.

## 🎚️ Recursos e personalização

Tudo é controlável direto na barra lateral:

| Configuração | O que faz |
| --- | --- |
| 🗂️ **Base de Dados** | Escolhe a base de conhecimento usada como contexto (ex.: *Oboticário*). |
| 🎯 **Objetivo do Conteúdo** | Define a meta do texto (ex.: *artigos de blog para atrair leads no topo do funil, com tom informativo*). |
| ➕ **Requisitos Extras** | Instruções e restrições específicas (ex.: *evitar jargões, incluir CTA, usar subtítulos, não citar concorrentes*). |
| 🔢 **Quantidade de conteúdo** | Número de posts gerados por pedido (1–5). |
| 📚 **Máximo de resultados (contexto)** | Quantos trechos da base alimentam a geração (profundidade do RAG). |
| 📏 **Faixa de tamanho** | Intervalo de comprimento do texto (ex.: 700–1200). |
| 🤖 **Modelo de Linguagem (LLM)** | Seleção do modelo (ex.: `gpt-4.1-mini`). |
| 🌐 **Idioma** | Idioma de saída do conteúdo. |

 É essa combinação de controles que dá aos posts uma **alta personalização** — do público-alvo ao tamanho, do tom ao modelo usado.
