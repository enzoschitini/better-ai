# Launch API — Ambiente de Testes de Endpoints

## Visão Geral

O **Launch API** é uma API leve construída com FastAPI, criada com o objetivo de:

> **Testar endpoints de forma rápida e isolada antes de integrá-los à API principal**
> 

Ele funciona como um ambiente de validação, permitindo que você desenvolva, experimente e debugue funcionalidades sem impactar o sistema principal.

---

## Objetivos

- Testar endpoints individualmente
- Validar lógica de negócio antes da produção
- Simular comportamentos da API principal
- Acelerar o desenvolvimento e debugging
- Reduzir risco de bugs em produção

---

### Objetivo:

---

## Como Executar

### Comando padrão:

```bash
uvicorn app:app --reload
```

ou

```bash
uvicorn local.launch_api:app --reload
```

### Explicação:

- `-reload` → reinicia automaticamente ao salvar alterações
- Ideal para desenvolvimento

---

## Casos de Uso

Este Launch API deve ser utilizado quando você quiser:

- Criar um endpoint novo antes de levar para produção
- Testar integração com:
    - Agents
    - Tools
    - RAG
- Validar entrada/saída de dados
- Simular comportamento de serviços externos

---

## Fluxo de Desenvolvimento Recomendado

1. Criar endpoint no **Launch API**
2. Testar isoladamente (Postman, curl, frontend)
3. Validar lógica e resposta
4. Refatorar (se necessário)
5. Migrar para API principal

---

## Boas Práticas

- ❌ Não usar dados sensíveis reais
- ❌ Não expor essa API publicamente
- ✅ Manter simples e focada em testes
- ✅ Criar endpoints pequenos e específicos
- ✅ Remover endpoints experimentais após uso
