# Application Setup and Initialization Guide

This guide provides a step-by-step process to configure your local environment and run the application.

---

## 1. Update the Project

First, make sure you are on the correct branch and that your code is up to date:

```bash
git checkout production
git pull
git checkout -b your-branch-name
```

---

## 2. Create the Virtual Environment

Create a virtual environment to isolate the project dependencies:

```bash
python -m venv .venv
```

---

## 3. Activate the Virtual Environment

### PowerShell

```bash
.\.venv\Scripts\Activate.ps1
```

### CMD

```bash
.venv\Scripts\activate.bat
```

After activation, your terminal should look something like this:

```bash
(.venv) PS C:\Users\user_name\better-ai>
```

---

## 4. Install Dependencies

You can install dependencies in two ways:

### Using uv (recommended)

```bash
uv sync
```

### Using pip

```bash
pip install -r requirements.txt
```

---

## 5. Run the API

Start the local server using Uvicorn:

```bash
uvicorn app:app --reload
```

---

## 7. Test the Application

Verify that the API is working properly using the health check endpoint:

```bash
curl --location 'http://localhost:8000/healthy'
```

If everything is correct, you should receive a response indicating that the application is healthy.

---

## Notes

- Always activate the virtual environment before running the application
- Prefer using `uv sync` to ensure dependency consistency
- If you encounter issues, recreate the virtual environment:

    ```bash
    rm -rf .venv
    python -m venv .venv
    ```

    ```bash
    deactivate
    Remove-Item -Recurse -Force .\venv
    ```
