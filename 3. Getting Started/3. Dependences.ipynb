# 📦 Documentação de Dependências

Este projeto utiliza um conjunto robusto de bibliotecas para construção de APIs, orquestração de agentes de IA, processamento de dados, armazenamento e integração com serviços externos.

As dependências estão organizadas por responsabilidade para facilitar manutenção, escalabilidade e entendimento da arquitetura.

---

## 1. Core & API Infrastructure

Responsáveis por estruturar a aplicação backend e comunicação HTTP.

| Biblioteca | Versão | Descrição |
| --- | --- | --- |
| dotenv | ~0.9.9 | Gerenciamento de variáveis de ambiente via `.env` |
| fastapi | ~0.120.4 | Framework moderno para criação de APIs rápidas em Python |
| uvicorn | ~0.23.2 | Servidor ASGI para rodar aplicações FastAPI |
| python-multipart | ~0.0.20 | Suporte a upload de arquivos em APIs |
| requests | ~2.32.5 | Cliente HTTP para consumir APIs externas |
