```python
class Produto:
    """
    Classe que representa um produto de estoque.
    """

    # Atributo de classe (compartilhado entre todas as instâncias)
    taxa_imposto = 0.1

    def __init__(self, nome: str, preco: float, quantidade: int = 0):
        """
        Construtor da classe.
        """
        self.nome = nome
        self._preco = preco           # atributo "protegido"
        self.__quantidade = quantidade  # atributo "privado"

    # ======== PROPRIEDADES ========

    @property
    def preco(self) -> float:
        return self._preco

    @preco.setter
    def preco(self, novo_preco: float):
        if novo_preco < 0:
            raise ValueError("Preço não pode ser negativo.")
        self._preco = novo_preco

    @property
    def quantidade(self) -> int:
        return self.__quantidade

    # ======== MÉTODOS PÚBLICOS ========

    def valor_total(self) -> float:
        """
        Calcula o valor total do estoque do produto.
        """
        return self._preco * self.__quantidade

    def aplicar_imposto(self):
        """
        Aplica o imposto ao preço do produto.
        """
        self._preco += self._preco * Produto.taxa_imposto

    # ======== MÉTODO PRIVADO ========

    def __registrar_log(self, mensagem: str):
        print(f"[LOG] {mensagem}")

    # ======== MÉTODO DE CLASSE ========

    @classmethod
    def alterar_taxa_imposto(cls, nova_taxa: float):
        cls.taxa_imposto = nova_taxa

    # ======== MÉTODO ESTÁTICO ========

    @staticmethod
    def validar_nome(nome: str) -> bool:
        return len(nome.strip()) > 0

    # ======== REPRESENTAÇÃO DA INSTÂNCIA ========

    def __str__(self):
        return f"Produto(nome={self.nome}, preço={self._preco}, quantidade={self.__quantidade})"
```

```python
class Animal:
    def __init__(self, nome):
        self.nome = nome

    def emitir_som(self):
        return "Som desconhecido"

    def info(self):
        return f"Eu sou um animal chamado {self.nome}"

class Cachorro(Animal):
    def __init__(self, nome, raca):
        # Chamando o construtor da classe pai
        super().__init__(nome)
        self.raca = raca

    # Sobrescrevendo método herdado
    def emitir_som(self):
        return "Au au!"

    def info(self):
        return f"{self.nome} é um cachorro da raça {self.raca}"

animal = Animal("Genérico")
dog = Cachorro("Rex", "Labrador")

print(animal.emitir_som())   # Som desconhecido
print(dog.emitir_som())      # Au au!
print(dog.info())            # Rex é um cachorro da raça Labrador
```
