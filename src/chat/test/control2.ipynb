## Tracing Tokens

```python
def menage_cost(tokens_response: dict, model_name: str = "gpt-4o-mini") -> dict:
    """
    Calcula o custo total de input e output com base no modelo e nos tokens.
    Retorna os valores formatados com 6 casas decimais como string (sem notação científica).
    """
    if model_name not in COST_MODELS:
        raise ValueError(f"Modelo '{model_name}' não encontrado em COST_MODELS.")

    model_cost = COST_MODELS[model_name]

    total_input_tokens = tokens_response["tokens_estimados"]["input"]["combined"]["tokens_estimated"]
    total_output_tokens = tokens_response["tokens_estimados"]["output"]["tokens_estimated"]

    cost_input = (total_input_tokens / 1_000_000) * model_cost["input"]
    cost_output = (total_output_tokens / 1_000_000) * model_cost["output"]
    cost_total = cost_input + cost_output

    # formatar para string com 6 casas decimais, sem notação científica
    def format_decimal(value):
        return f"{value:.6f}"

    cost = {
        "modelo": model_name,

        "input_tokens": total_input_tokens,
        "output_tokens": total_output_tokens,
        "total_tokens": (total_input_tokens + total_output_tokens),
        
        "input_cost_usd": format_decimal(cost_input),
        "output_cost_usd": format_decimal(cost_output),
        "total_cost_usd": format_decimal(cost_total),
    }
    
    tokens_response["cost"] = cost

    return tokens_response

tokens_response_update = menage_cost(tokens_response=dic, model_name="gpt-4o-mini")
#print(json.dumps(tokens_response_update, indent=4))
```

## Verify

```python
# Verifica se supera la soglia del piano
def verificar_status_plano(dados: dict) -> str:
    """
    Verifica se o uso de tokens ou custos ultrapassa os limites do plano.
    Retorna:
        "no_activated" se algum limite for ultrapassado
        "activated" caso contrário
    """
    try:
        plan = dados["plan"]
        resource_tokens = plan["resorce"]["tokens"]
        resource_costs = plan["resorce"]["cost"]
        used_tokens = plan["tokens"]
        used_costs = plan["cost"]

        # --- Verifica limites de tokens ---
        if (used_tokens["input_tokens"] > resource_tokens["monthly_budget_input_tokens"] or
            used_tokens["output_tokens"] > resource_tokens["monthly_budget_output_tokens"] or
            used_tokens["total_tokens"] > resource_tokens["monthly_budget_total_tokens"]):
            return "no_activated"

        # --- Verifica limites de custo ---
        if (float(used_costs["input_cost_usd"]) > float(resource_costs["monthly_budget_input_cost_usd"]) or
            float(used_costs["output_cost_usd"]) > float(resource_costs["monthly_budget_output_cost_usd"]) or
            float(used_costs["total_cost_usd"]) > float(resource_costs["monthly_budget_total_cost_usd"])):
            return "no_activated"

        return "activated"

    except KeyError as e:
        raise ValueError(f"Campo ausente no JSON: {e}")

verificar_status_plano(plan_status)
```

```python
def update_status_plan(business_alvo, status_novo):
    # Caminho do arquivo
    caminho_arquivo = "business_acess.json"

    # 1️⃣ Carrega o JSON
    with open(caminho_arquivo, "r", encoding="utf-8") as arquivo:
        dados = json.load(arquivo)

    # 2️⃣ Atualiza apenas o item com o client_id desejado
    for item in dados:
        if item.get("client_id") == business_alvo:
            item["status_plan"] = status_novo

    # 3️⃣ Salva de volta no arquivo
    with open(caminho_arquivo, "w", encoding="utf-8") as arquivo:
        json.dump(dados, arquivo, indent=4, ensure_ascii=False)

update_status_plan(client_id, business_plan_status)
```
