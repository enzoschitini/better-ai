```python
import json
import pandas as pd
from io import BytesIO
```

```python
with open(f"supermarket_sales.csv", "rb") as f:
    file_bytes = f.read()

_df = pd.read_csv(BytesIO(file_bytes))
columns = _df.columns.tolist()
df = _df[columns[:50]]
```

```python
markdown_table = df.head(n=10).to_markdown(index=False)
print(markdown_table)
print(df.shape)
```

```python
columns = df.columns.tolist()
print(columns)
```

```python
for col in columns:
    print(f"Column: {col}")
    type_info = df[col].dtype
    print(f"Data type: {type_info}")
    print(f"Is null count: {df[col].isnull().sum()}")

    if type_info == "object":
        print(f"5 Unique Values: {df[col].unique()[:5]}")

    print(df[col].describe())
```
