## Introduction

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(filename)s - line: %(lineno)d - %(levelname)s - %(message)s'
)

logging.debug("Mensagem de debug")
logging.info("Mensagem informativa")
logging.warning("Aviso!")
logging.error("Erro encontrado!")
logging.critical("Erro muito grave!")
```

| Nível    | Descrição                                          |
| -------- | -------------------------------------------------- |
| DEBUG    | Informações detalhadas para desenvolvedores        |
| INFO     | Mensagens de fluxo normal do programa              |
| WARNING  | Algo inesperado, mas que o programa pode continuar |
| ERROR    | Erro que afetou alguma funcionalidade              |
| CRITICAL | Erro grave, programa pode parar                    |


### Salvando logs em arquivo

```python
import logging

logging.basicConfig(
    filename='app.log',
    level=logging.DEBUG,
    format='%(asctime)s - %(levelname)s - %(message)s'
)

logging.info("Aplicação iniciada")
logging.error("Erro simulado")
```

### *Formato da mensagem (format)*

| Placeholder     | Mostra              |
| --------------- | ------------------- |
| `%(asctime)s`   | Data/hora           |
| `%(levelname)s` | Nível (INFO, ERROR) |
| `%(message)s`   | Mensagem            |
| `%(name)s`      | Nome do logger      |
| `%(filename)s`  | Nome do arquivo     |

### Handlers múltiplos (console + arquivo)

```python
import logging

logger = logging.getLogger("meu_logger")
logger.setLevel(logging.DEBUG)

# handler para arquivo
file_handler = logging.FileHandler("app.log")
file_handler.setLevel(logging.DEBUG)

# handler para console
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)

# formatação
formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s')
file_handler.setFormatter(formatter)
console_handler.setFormatter(formatter)

logger.addHandler(file_handler)
logger.addHandler(console_handler)

logger.info("Mensagem de teste")
logger.debug("Debug em arquivo, mas não no console")
```
