# 📦 Documentação de Dependências

As dependências estão organizadas por responsabilidade para facilitar manutenção, escalabilidade e entendimento da arquitetura.

**Projeto:** `better-ai` · **Versão:** `0.1.0` · **Python requerido:** `>=3.12`

> Fonte da verdade: `pyproject.toml`. Instalação recomendada com `uv sync` (ou `uv pip install -e .`).

## Convenções de versionamento

| Operador | Significado | Exemplo |
| --- | --- | --- |
| `~=X.Y.Z` | Permite atualizações de patch, fixa major e minor | `~=0.3.27` → `>=0.3.27, <0.4.0` |
| `>=X.Y.Z` | Aceita qualquer versão igual ou superior | `>=1.56.0` |

As bibliotecas com `~=` são as que impactam diretamente o comportamento dos agentes e dos pipelines de RAG — mantê-las fixadas por minor evita quebras silenciosas de API. As marcadas com `>=` são periféricas ou estáveis o suficiente para acompanhar as versões mais recentes.
